```python
from dotenv import load_dotenv
from core.validation import get_valid_image_number
import cv2 as cv
from numpy import ndarray
import numpy as np
import os


### Setup ###
# Load variables from .env file
load_dotenv()

# Base Path for image folder
base_mask_folder: str = os.getenv("BASE_MASK_FOLDER")
output_folder: str = os.getenv("OUTPUT_FOLDER")

img_number: str = get_valid_image_number()
mask_path: str = os.path.join(base_mask_folder, f"{img_number}.png")

# Load and Convert Images
bgr_img = cv.imread(mask_path, 1)
hsv_img = cv.cvtColor(bgr_img, cv.COLOR_BGR2HSV)

# Print image dimensions for verification
print(f"Image dimensions (height, width): {hsv_img.shape[:2]}")

# Define color limits for all 18 elements
color_limits = [
    # Aqua Color
    (np.array([70, 82, 255]), np.array([90, 255, 255])),
    # Aquamarine Color
    (np.array([78, 129, 241]), np.array([79, 145, 246])),
    # Blue Color
    (np.array([120, 247, 255]), np.array([120, 255, 255])),
    # Fuchsia Color
    (np.array([150, 255, 255]), np.array([150, 255, 255])),
    # Green Color
    (np.array([59, 248, 255]), np.array([60, 255, 255])),
    # Light Amber Color
    (np.array([19, 91, 241]), np.array([20, 94, 242])),
    # Light Blue Violet Color
    (np.array([115, 117, 235]), np.array([115, 124, 237])),
    # Light Blue Color
    (np.array([106, 71, 235]), np.array([107, 73, 236])),
    # Light Coral Color
    (np.array([6, 23, 225]), np.array([9, 42, 235])),
    # Light Gray Color
    (np.array([27, 5, 203]), np.array([38, 12, 206])),
    # Light Khaki Color
    (np.array([26, 80, 251]), np.array([26, 88, 252])),
    # Light Olive Color
    (np.array([37, 79, 217]), np.array([47, 130, 235])),
    # Light Orange Color
    (np.array([12, 61, 230]), np.array([15, 95, 236])),
    # Light Teal Color
    (np.array([73, 108, 254]), np.array([74, 111, 255])),
    # Medium Purple Color
    (np.array([138, 59, 176]), np.array([139, 81, 187])),
    # Pale Aqua Color
    (np.array([77, 51, 253]), np.array([86, 60, 254])),
    # Red Color
    (np.array([0, 240, 255]), np.array([0, 255, 255])),
    # Yellow Color
    (np.array([30, 242, 255]), np.array([30, 255, 255]))
]

# Create masks for each color and concatenate them
masks: list[ndarray] = []
for lower_bound, upper_bound in color_limits:
    mask: ndarray = cv.inRange(hsv_img, lower_bound, upper_bound)
    # Add new axis for each mask binary mask
    mask = mask[..., np.newaxis]
    masks.append(mask)

# Concatenate masks along the channel dimension Z
combined_mask: ndarray = np.concatenate(masks, axis=2)

# Save combined mask as a NumPy file
output_file: str = os.path.join(output_folder, f"{img_number}.npy")
np.save(output_file, combined_mask)
```
