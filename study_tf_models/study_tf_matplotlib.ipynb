```python
# Adicionar no início do script, antes dos imports de plotting
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')  # Backend sem interface gráfica
plt.rcParams['figure.dpi'] = 300
plt.rcParams['savefig.dpi'] = 300

print("Backend atual:", plt.get_backend())
print("DPI padrão da figura:", plt.rcParams['figure.dpi'])
print("DPI padrão para salvar:", plt.rcParams['savefig.dpi'])
```

```python
# Gerar uma imagem de teste
import matplotlib.pyplot as plt
matplotlib.use('Agg')  # Backend sem interface gráfica
plt.rcParams['figure.dpi'] = 300
plt.rcParams['savefig.dpi'] = 300

fig, ax = plt.subplots(figsize=(4, 3))
ax.plot([1, 2, 3], [1, 4, 2])
ax.set_title('Teste 300 DPI')
fig.savefig('teste_300dpi.png', dpi=300, bbox_inches='tight')
plt.close()
```
