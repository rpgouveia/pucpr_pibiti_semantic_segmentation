```python
import cv2
import numpy as np

def nothing(x):
    pass

def create_hsv_tuner(image_path):
    # Criar janela para os trackbars
    cv2.namedWindow('HSV Tuner')
    
    # Criar trackbars para primeiro intervalo
    cv2.createTrackbar('H min 1', 'HSV Tuner', 0, 180, nothing)
    cv2.createTrackbar('H max 1', 'HSV Tuner', 10, 180, nothing)
    cv2.createTrackbar('S min 1', 'HSV Tuner', 100, 255, nothing)
    cv2.createTrackbar('S max 1', 'HSV Tuner', 255, 255, nothing)
    cv2.createTrackbar('V min 1', 'HSV Tuner', 255, 255, nothing)
    cv2.createTrackbar('V max 1', 'HSV Tuner', 255, 255, nothing)
    
    # Criar trackbars para segundo intervalo
    cv2.createTrackbar('H min 2', 'HSV Tuner', 160, 180, nothing)
    cv2.createTrackbar('H max 2', 'HSV Tuner', 180, 180, nothing)
    cv2.createTrackbar('S min 2', 'HSV Tuner', 240, 255, nothing)
    cv2.createTrackbar('S max 2', 'HSV Tuner', 255, 255, nothing)
    cv2.createTrackbar('V min 2', 'HSV Tuner', 0, 255, nothing)
    cv2.createTrackbar('V max 2', 'HSV Tuner', 255, 255, nothing)
    
    # Criar trackbar para operações morfológicas
    cv2.createTrackbar('Kernel Size', 'HSV Tuner', 0, 20, nothing)
    
    # Ler a imagem
    image = cv2.imread(image_path)
    
    # Redimensionar se a imagem for muito grande
    max_dimension = 512
    height, width = image.shape[:2]
    if max(height, width) > max_dimension:
        scale = max_dimension / max(height, width)
        image = cv2.resize(image, None, fx=scale, fy=scale)
    
    while True:
        # Converter para HSV
        hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
        
        # Obter valores atuais dos trackbars - Primeiro intervalo
        h_min_1 = cv2.getTrackbarPos('H min 1', 'HSV Tuner')
        h_max_1 = cv2.getTrackbarPos('H max 1', 'HSV Tuner')
        s_min_1 = cv2.getTrackbarPos('S min 1', 'HSV Tuner')
        s_max_1 = cv2.getTrackbarPos('S max 1', 'HSV Tuner')
        v_min_1 = cv2.getTrackbarPos('V min 1', 'HSV Tuner')
        v_max_1 = cv2.getTrackbarPos('V max 1', 'HSV Tuner')
        
        # Obter valores atuais dos trackbars - Segundo intervalo
        h_min_2 = cv2.getTrackbarPos('H min 2', 'HSV Tuner')
        h_max_2 = cv2.getTrackbarPos('H max 2', 'HSV Tuner')
        s_min_2 = cv2.getTrackbarPos('S min 2', 'HSV Tuner')
        s_max_2 = cv2.getTrackbarPos('S max 2', 'HSV Tuner')
        v_min_2 = cv2.getTrackbarPos('V min 2', 'HSV Tuner')
        v_max_2 = cv2.getTrackbarPos('V max 2', 'HSV Tuner')
        
        # Obter tamanho do kernel
        kernel_size = cv2.getTrackbarPos('Kernel Size', 'HSV Tuner')
        kernel_size = max(1, kernel_size) # Garantir que seja pelo menos 1
        
        # Definir intervalos
        lower_1 = np.array([h_min_1, s_min_1, v_min_1])
        upper_1 = np.array([h_max_1, s_max_1, v_max_1])
        lower_2 = np.array([h_min_2, s_min_2, v_min_2])
        upper_2 = np.array([h_max_2, s_max_2, v_max_2])
        
        # Criar máscaras
        mask1 = cv2.inRange(hsv, lower_1, upper_1)
        mask2 = cv2.inRange(hsv, lower_2, upper_2)
        mask = cv2.bitwise_or(mask1, mask2)
        
        # Aplicar operações morfológicas
        kernel = np.ones((kernel_size, kernel_size), np.uint8)
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
        
        # Aplicar máscara na imagem
        result = cv2.bitwise_and(image, image, mask=mask)
        
        # Mostrar imagens
        cv2.imshow('Original', image)
        cv2.imshow('Mask', mask)
        cv2.imshow('Result', result)
        
        # Mostrar valores atuais
        print('\nIntervalo 1:', lower_1, upper_1)
        print('Intervalo 2:', lower_2, upper_2)
        print('Kernel Size:', kernel_size)
        
        # Esperar por tecla
        key = cv2.waitKey(1) & 0xFF
        if key == ord('q'):
            break
        elif key == ord('s'):
            # Salvar valores em um arquivo
            values = {
                'interval_1': {'lower': lower_1.tolist(), 'upper': upper_1.tolist()},
                'interval_2': {'lower': lower_2.tolist(), 'upper': upper_2.tolist()},
                'kernel_size': kernel_size
            }
            np.save('hsv_values.npy', values)
            print('\nValores salvos em hsv_values.npy')
    
    cv2.destroyAllWindows()

if __name__ == "__main__":
    image_path = "/media/renato/Data/PIBIC/FinalImagePPM_Mask_v2/000.png"
    create_hsv_tuner(image_path)
```

```python
import cv2
import numpy as np

def nothing(x):
    pass

def create_dark_green_tuner(image_path):
    # Criar janela para os trackbars
    cv2.namedWindow('Dark Green Tuner')
    
    # Criar trackbars com valores iniciais para o verde escuro
    cv2.createTrackbar('H min', 'Dark Green Tuner', 45, 180, nothing)
    cv2.createTrackbar('H max', 'Dark Green Tuner', 60, 180, nothing)
    cv2.createTrackbar('S min', 'Dark Green Tuner', 160, 255, nothing)
    cv2.createTrackbar('S max', 'Dark Green Tuner', 255, 255, nothing)
    cv2.createTrackbar('V min', 'Dark Green Tuner', 50, 255, nothing)
    cv2.createTrackbar('V max', 'Dark Green Tuner', 130, 255, nothing)
    cv2.createTrackbar('Kernel Size', 'Dark Green Tuner', 5, 20, nothing)
    
    # Ler a imagem
    image = cv2.imread(image_path)
    
    # Redimensionar se a imagem for muito grande
    max_dimension = 800
    height, width = image.shape[:2]
    if max(height, width) > max_dimension:
        scale = max_dimension / max(height, width)
        image = cv2.resize(image, None, fx=scale, fy=scale)
    
    while True:
        # Converter para HSV
        hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
        
        # Obter valores atuais dos trackbars
        h_min = cv2.getTrackbarPos('H min', 'Dark Green Tuner')
        h_max = cv2.getTrackbarPos('H max', 'Dark Green Tuner')
        s_min = cv2.getTrackbarPos('S min', 'Dark Green Tuner')
        s_max = cv2.getTrackbarPos('S max', 'Dark Green Tuner')
        v_min = cv2.getTrackbarPos('V min', 'Dark Green Tuner')
        v_max = cv2.getTrackbarPos('V max', 'Dark Green Tuner')
        kernel_size = cv2.getTrackbarPos('Kernel Size', 'Dark Green Tuner')
        
        # Definir intervalos
        lower = np.array([h_min, s_min, v_min])
        upper = np.array([h_max, s_max, v_max])
        
        # Criar máscara
        mask = cv2.inRange(hsv, lower, upper)
        
        # Aplicar operações morfológicas
        kernel = np.ones((max(1, kernel_size), max(1, kernel_size)), np.uint8)
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
        mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
        
        # Aplicar máscara na imagem
        result = cv2.bitwise_and(image, image, mask=mask)
        
        # Mostrar imagens
        cv2.imshow('Original', image)
        cv2.imshow('Mask', mask)
        cv2.imshow('Result', result)
        
        # Mostrar valores atuais
        print('\nHSV Range:', lower, upper)
        print('Kernel Size:', kernel_size)
        
        # Esperar por tecla
        key = cv2.waitKey(1) & 0xFF
        if key == ord('q'):
            break
        elif key == ord('s'):
            # Salvar valores em um arquivo
            values = {
                'lower': lower.tolist(),
                'upper': upper.tolist(),
                'kernel_size': kernel_size
            }
            np.save('dark_green_values.npy', values)
            print('\nValores salvos em dark_green_values.npy')
    
    cv2.destroyAllWindows()

if __name__ == "__main__":
    image_path = "/media/renato/Data/PIBIC/FinalImagePPM_Mask_v2/2635.png"
    create_dark_green_tuner(image_path)
```
